# Парабола и квадратичная функция (8 класс)

Этот урок помогает понять, как работает функция вида:

$$y = ax^2 + bx + c$$

и как решать квадратичные неравенства через график.

## Что мы изучим

1. Что означает каждый коэффициент `a`, `b`, `c`.
2. Как меняется график (парабола) при изменении коэффициентов.
3. Как решать квадратичные неравенства графически.
4. Небольшие тренировочные задачи.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (8, 5)
plt.rcParams['axes.grid'] = True

def f(x, a, b, c):
    return a * x**2 + b * x + c

x = np.linspace(-8, 8, 400)
```

## Базовая теория

Для функции `y = ax^2 + bx + c`:

- `a` влияет на направление ветвей и "ширину" параболы.
- `b` влияет на положение вершины по горизонтали.
- `c` показывает, где график пересекает ось `Oy` (`y(0)=c`).

Вершина параболы имеет абсциссу:

$$x_{верш} = -\frac{b}{2a}$$

Если `a > 0`, ветви вверх. Если `a < 0`, ветви вниз.

## Влияние коэффициента `a`

Сравним несколько парабол при одинаковых `b=0`, `c=0`.

```python
a_values = [2, 1, 0.5, -1, -2]

plt.figure(figsize=(9, 6))
for a in a_values:
    y = f(x, a, 0, 0)
    plt.plot(x, y, label=f'a = {a}')

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.ylim(-20, 20)
plt.title('Как коэффициент a меняет параболу')
plt.legend()
plt.show()
```

Вывод:

- `a > 0` -> ветви вверх.
- `a < 0` -> ветви вниз.
- Чем больше `|a|`, тем парабола уже (круче).
- Чем меньше `|a|` (но не 0), тем парабола шире.

## Влияние коэффициента `b`

Пусть `a=1`, `c=0`. Меняем только `b`.

```python
b_values = [-6, -2, 0, 2, 6]

plt.figure(figsize=(9, 6))
for b in b_values:
    y = f(x, 1, b, 0)
    x_vertex = -b / 2
    y_vertex = f(x_vertex, 1, b, 0)
    plt.plot(x, y, label=f'b = {b}, x_верш = {x_vertex:.1f}')
    plt.scatter([x_vertex], [y_vertex], s=25)

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.ylim(-15, 35)
plt.title('Как коэффициент b сдвигает вершину по x')
plt.legend()
plt.show()
```

Вывод:

- `b` меняет положение вершины по оси `x`.
- Формула вершины по `x`: `x = -b/(2a)`.

## Влияние коэффициента `c`

Пусть `a=1`, `b=0`. Меняем только `c`.

```python
c_values = [-6, -2, 0, 3, 7]

plt.figure(figsize=(9, 6))
for c in c_values:
    y = f(x, 1, 0, c)
    plt.plot(x, y, label=f'c = {c}')
    plt.scatter([0], [c], s=30)

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.ylim(-10, 25)
plt.title('Как коэффициент c сдвигает параболу вверх/вниз')
plt.legend()
plt.show()
```

Вывод:

- `c` отвечает за пересечение с осью `Oy`: точка `(0, c)`.
- Изменение `c` сдвигает весь график вверх или вниз.

## Квадратичные неравенства: графическая идея

Пример: решить неравенство

$$x^2 - 3x - 4 > 0$$

Шаги:
1. Строим график `y = x^2 - 3x - 4`.
2. Находим, где график выше оси `Ox` (`y>0`) или ниже (`y<0`).
3. По оси `x` записываем промежутки.

```python
def plot_inequality(a, b, c, sign='>0', xlim=(-8, 8), ylim=(-15, 20), title=''):
    x_local = np.linspace(xlim[0], xlim[1], 600)
    y_local = f(x_local, a, b, c)

    plt.figure(figsize=(9, 5))
    plt.plot(x_local, y_local, label=f'y={a}x^2+{b}x+{c}', color='royalblue')
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1)

    if sign == '>0':
        mask = y_local > 0
        label = 'Область, где y > 0'
    elif sign == '<0':
        mask = y_local < 0
        label = 'Область, где y < 0'
    elif sign == '>=0':
        mask = y_local >= 0
        label = 'Область, где y >= 0'
    else:
        mask = y_local <= 0
        label = 'Область, где y <= 0'

    plt.fill_between(x_local, y_local, 0, where=mask, alpha=0.3, color='orange', label=label)

    # Корни (если есть) через дискриминант
    D = b**2 - 4*a*c
    if D > 0:
        x1 = (-b - np.sqrt(D)) / (2*a)
        x2 = (-b + np.sqrt(D)) / (2*a)
        plt.scatter([x1, x2], [0, 0], color='red', zorder=5)
        plt.text(x1, 0.8, f'x1={x1:.2f}', color='red')
        plt.text(x2, 0.8, f'x2={x2:.2f}', color='red')
    elif D == 0:
        x0 = -b / (2*a)
        plt.scatter([x0], [0], color='red', zorder=5)
        plt.text(x0, 0.8, f'x0={x0:.2f}', color='red')

    plt.xlim(*xlim)
    plt.ylim(*ylim)
    plt.title(title)
    plt.legend()
    plt.show()

plot_inequality(1, -3, -4, sign='>0', title='Пример 1: x^2 - 3x - 4 > 0')
```

Для примера выше корни: `x=-1` и `x=4`.

Так как ветви вверх, `y>0` вне корней:

$$x \in (-\infty, -1) \cup (4, +\infty)$$

## Еще примеры неравенств

Ниже три задачи с графической интерпретацией.

```python
plot_inequality(1, -2, -3, sign='<0', title='Задача 1: x^2 - 2x - 3 < 0')
plot_inequality(-1, -2, 3, sign='>=0', title='Задача 2: -x^2 - 2x + 3 >= 0')
plot_inequality(2, 4, 2, sign='<=0', title='Задача 3: 2x^2 + 4x + 2 <= 0')
```
