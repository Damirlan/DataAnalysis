# Лабораторная 2, Задача 2: Проверка надёжности пароля

**Что требуется сделать:**

Нужно проверить список паролей на соответствие правилам сложности. Для каждого пароля программа должна определить, валидный он или нет.

**Критерии валидности пароля:**

- длина не меньше 12 символов;
- есть хотя бы одна заглавная буква;
- есть хотя бы одна строчная буква;
- есть хотя бы одна цифра;
- есть хотя бы один специальный символ из заданного набора.

**Как я решал:**

Я вынес проверку в отдельную функцию `is_valid_password(password)`, чтобы код был переиспользуемым и читаемым. Для поиска нужных типов символов использую регулярные выражения (`re.search`). В конце собираю все проверки в `all(...)`.

**Алгоритм по шагам:**

1. Сначала отсекаю слишком короткие пароли.
2. Проверяю наличие каждого обязательного типа символов.
3. Возвращаю `True`, только если все условия выполнены одновременно.
4. Считываю `N` и последовательно обрабатываю `N` паролей.

**Сложность:**

Для одного пароля длины `L` время проверки линейное, то есть `O(L)`.


```python
import re

def is_valid_password(password):
    # Проверка минимальной длины
    if len(password) < 12:
        return False
    
    # Регулярные выражения для проверки наличия необходимых символов
    has_uppercase = bool(re.search(r'[A-Z]', password))   # Заглавная буква
    has_lowercase = bool(re.search(r'[a-z]', password))   # Строчная буква
    has_digit = bool(re.search(r'\d', password))          # Цифра
    has_special_char = bool(re.search(r'[!\@\#\$\%\&\*\+]', password))  # Спецсимвол
    
    # Все условия должны выполняться одновременно
    return all([has_uppercase, has_lowercase, has_digit, has_special_char])

# Чтение входных данных
N = int(input("Количество паролей: "))
passwords = [input() for _ in range(N)]

# Проверка всех паролей и вывод результатов
for pwd in passwords:
    print("Valid" if is_valid_password(pwd) else "Invalid")
```
