# Лабораторная 2, Задача 1: Сжатие строки

**Что требуется сделать:**

В этой задаче нужно реализовать сжатие строки по группам одинаковых соседних символов. Если символ повторяется несколько раз подряд, я записываю сначала количество повторов, потом сам символ. Если символ встретился один раз, записываю только символ.

**Как я понял идею решения:**

Я прохожу по строке слева направо и считаю длину текущей серии одинаковых символов. Когда символ меняется, добавляю в результат накопленную серию и начинаю считать новую.

**Алгоритм по шагам:**

1. Проверяю крайний случай: пустая строка или строка длины 1.
2. Инициализирую счётчик текущей серии (`count = 1`).
3. Иду циклом по строке со второго символа.
4. Если символ совпадает с предыдущим, увеличиваю счётчик.
5. Если не совпадает, добавляю серию в результат и сбрасываю счётчик.
6. После цикла обязательно добавляю последнюю серию.

**Почему это решение корректно:**

Каждый символ обрабатывается ровно один раз, и каждая серия фиксируется либо при смене символа, либо в конце строки.

**Сложность:**

- Время: `O(n)`
- Память: `O(n)` на результирующую строку


```python
# Ячейка 1: Определение функции сжатия
def compress_string(s):
    if len(s) <= 1:
        return s
    
    result = []
    count = 1
    
    for i in range(1, len(s)):
        if s[i] != s[i-1]:
            if count > 1:
                result.append(str(count))
            result.append(s[i-1])
            count = 1
        else:
            count += 1
            
    if count > 1:
        result.append(str(count))
    result.append(s[-1])
    
    return ''.join(result)
```

```python
# Ячейка 2: Пример ввода и вывода
input_str = 'AAACBBBBBDDADDD'  # Исходная строка
output_str = compress_string(input_str)  # Полученный результат
print(f"Входные данные:\n{input_str}\n")
print(f"Выходные данные:\n{output_str}")
```
